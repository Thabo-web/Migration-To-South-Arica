```python

import pandas as pd
from pathlib import Path

try:
    from IPython.display import display
except ImportError:
    def display(obj):
        print(obj)

try:
    BASE = Path(__file__).resolve().parents[1]
except NameError:
    BASE = Path.cwd()

RAW_CSV = BASE / "data" / "API_SM.POP.NETM_DS2_en_csv_v2_34232.csv"

print("Reading:", RAW_CSV.resolve())

if not RAW_CSV.exists():
    raise FileNotFoundError(f"CSV not found at: {RAW_CSV}")

try:
    df = pd.read_csv(RAW_CSV, skiprows=4, encoding="utf-8-sig", low_memory=False)
    print("Rows:", len(df))
    display(df.head())
except Exception as e:
    print("read error:", e)
```
